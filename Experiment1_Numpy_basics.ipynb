```python
# Read bytes as integers
np.fromiter(a, dtype=np.uint8)  # Output: array of ASCII values

# Read bytes as characters (1-byte strings)
np.fromiter((bytes([x]) for x in a), dtype="S1")  # Output: array([b'h', b'e', ...])
```

** 1. NumPy Fundamentals: Array creation: 1D, 2D, 3D arrays, Indexing, slicing, broadcasting,
Arithmetic operations and matrix computations **

```python
# Indexing
# accessing elements in a 1D array
print(array_1d[0])
```

```python
# accessing elements in a 2D array
print(array_2d[0][1]) # to print element 2 from the array_2d
```

```python
# accessing elements in a 3D array
print(array_3d[1][0][1]) # to print 8 from the array_3d
```

```python
# slicing
print(array_2d[0:1])
```

```python

# Broadcasting
# Broadcasting allows NumPy to perform operations on arrays of different shapes.
# creating an array and a scalar
array_broadcast = np.array([1,2,3])
print(array_broadcast)
scalar = 2
# # multiplying the array by the scalar
result = array_broadcast * scalar
print(result)
```

```python
# arthimetic operations
# element wise addition
result_add = array_2d + 10
print(result_add)
```

```python
# element wise multiplication
result_mul = array_2d * 2
print(result_mul)
```

```python
# matrix computations
array_a = np.array([[1,2],[3,4]])
array_b = np.array([[5,6],[7,8]])
# matrix multiplication
result_matrix = np.dot(array_a,array_b)
print(result_matrix)
```

```python
# Matrix inversion
matrix = np.array([[1,2],[3,4]])
inverse_matrix = np.linalg.inv(matrix)
print(inverse_matrix)
```

```python
# determinant
determinant = np.linalg.det(matrix)
print(determinant)
```
